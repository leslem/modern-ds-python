Uncomment each one, observe the error, and fix it.

Snippet 2:<br>
<br>

## Exercise 6: String expressions

The code below shows the unique metric names from the GitHub source.
Names like `daily_issues_opened` are hard to read.
Transform them so that underscores are replaced with spaces and the result is title-cased (e.g., "Daily Issues Opened").

```python
github_metrics = metrics.filter(pl.col("source") == "github")
github_metrics.select("metric").unique().sort("metric")
```

**Optional extension:**

* Split each metric name into two new columns: `frequency` (the first word, e.g., "daily" or "total") and `measurement` (the rest, e.g., "issues opened").

**Hints** *(highlight to reveal)*

<div style="color: #555; background-color: #555; padding: 8px; border-radius: 4px;">
&bull; Use the <code>.str</code> namespace: <code>.str.replace_all()</code> and <code>.str.to_titlecase()</code>.<br>
&bull; For the extension, <code>.str.split("_")</code> returns a list column. Use <code>.list.first()</code> to get the first element.<br>
&bull; To get the remaining words, use <code>.list.slice(1)</code> followed by <code>.list.join(" ")</code>.<br>
</div>

**Answer** *(highlight to reveal)*

<div style="color: #555; background-color: #555; padding: 8px; border-radius: 4px;">
<pre style="color: inherit; background: inherit; margin: 4px 0; padding: 0;">(
    github_metrics
    .select("metric")
    .unique()
    .with_columns(
        clean_name=pl.col("metric").str.replace_all("_", " ").str.to_titlecase()
    )
    .sort("metric")
)</pre>
Extension:<br>
<br>
<pre style="color: inherit; background: inherit; margin: 4px 0; padding: 0;">(
    github_metrics
    .select("metric")
    .unique()

## Exercise 9: Joining DataFrames

The two DataFrames below summarize PyPI downloads and GitHub stars per project.
Join them on `project` so that you can see each project's downloads alongside its GitHub stars.

```python
pypi_totals = (
    downloads
    .group_by("project")
    .agg(total_downloads=pl.col("value").sum())
)
pypi_totals
```

```python
github_stars = (
    metrics.filter(
        pl.col("source") == "github",
        pl.col("metric") == "total_reactions",
    )
    .group_by("project")
    .agg(total_stars=pl.col("value").max())
)
github_stars
```

**Optional extension:**

## Exercise 10: Putting it all together

Answer this question using a single method chain:

Start from the full `metrics` DataFrame and use filter, group_by, agg, and sort.
